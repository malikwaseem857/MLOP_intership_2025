```python
## Simple Calculator with the help of nested if-else statement

num_1 = int(input("Enter a number:"))
num_2 = int(input("Enter the second number:"))
operation = input("Enter an operation(+, -, *, /,//,%):")

if operation == '+':
    sum = num_1 + num_2
    print(sum)
elif operation == '-':
    diff = num_1 - num_2
    print(diff)
elif operation == '*':
    multi = num_1 * num_2
    print(multi)
elif operation == '/':
    if num_2 != 0:
        div = num_1 / num_2
        print(div)
    else:
        print("Division by zero is not possible. Change the number.")
elif operation == '//':
    floor_div = num_1 // num_2
    print(floor_div)
elif operation == '%':
    modulo = num_1 % num_2
    print(modulo)
else:
    print("Invalid operation")
    

```

```python
## nested if-else

year = int(input("Enter the year:"))

if year%4== 0:
    if year%100==0:
        if year%400==0:
            print(f"The {year} is a leap year.")
        else:
            print(f"The {year} is not a leap year.")
        print(f'The {year} is not a leap year.')
    print(f"The {year} is a leap year.")
else:
    print(f"The {year} is not a leap year.")
```
