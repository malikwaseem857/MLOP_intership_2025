## Exception Handling in Python
- Try and Excpet Blocks

```python
try:
    a=b
except NameError as ex:
    print("Variable 'b' is not defined")
```

```python
## NameError
a=b
```

```python
result = 1/0
```

```python
try:
    result = 1/0
except ZeroDivisionError as ex:
    print(f"An error occurred: {ex}")
```

```python
## a=b is managed but not defined
try:
    result = 1/0
    a=b
except ZeroDivisionError as ex:
    print(f"An error occurred: {ex}")
```

```python
## now both are defined and managed
try:
    result = 1/0
    a=b
except ZeroDivisionError as ex:
    print(f"An error occurred: {ex}")

except NameError as ex1:
    print(f"An error occurred: {ex1}")
```
