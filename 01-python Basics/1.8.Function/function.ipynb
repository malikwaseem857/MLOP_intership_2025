```python
## if agruments are in key values 

def print_dict(**kargs):
    for key, value in kargs.items():
        print(f"{key}: {value}")
print_dict(name= "Waseem", age= 27, gender= "Male")
```

```python
def print_gen(*args, **kwargs):
    for arg in args:
        print(arg, end=' ')

    for key,value in kwargs.items():
        print(f'{key}, {value}', end=' ')


print_gen(1,2,3,4,5,'Waseem', 'Lahore', 'PU', degree = "Botany")
```

```python
def cube(num):
    if num>5:
        return num ** 3
```

```python
## map function work like for loop and apply function 
num2 = [4,5,6,7,6,10,0,2,5,7,8]

cube_result = list(map(cube, num2))

print(cube_result)
```

## Filter Function

```python
# it pply condition in each function

def even(num):
    if num % 2 == 0:
        return True
    
num2 = [4,5,6,7,6,10,0,2,5,7,8]
 
filtered_numbers = list(filter(even, num2))

print(filtered_numbers)
```
