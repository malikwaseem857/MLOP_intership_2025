## Arithematic Operators

```python
## +
## -
## /
## //
## %
 
div = 3/4  # simple division
floor_div = 3//4  # output convert lower neartest 0
modulo = 4%2  # means remainder 

print(div)
print(floor_div)
print(modulo)
```

```python
## Logical Operatoes

## and (Both are True or fullfilled)
## or  (Either one is True)
## not (Opposite of the statement)

x = True
y = True
print(x and y)

x = True
y = False
print(x and y)

x = True
y = True
print(not(x and y))
```
