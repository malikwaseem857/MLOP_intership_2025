## Ploymorphism in Python
- Method Overriding
- Abstruct Based Classes

```python
## Polymorphism
#Polymorphism is the ability of an object to take on multiple forms.

## our ultimate goal is speak this animal

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print('The animal makes a sound')

class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)

    def speak(self):
        print(f'The {self.name} says woof!')  # method same task different

# Function to demonstrate polymorphism
def animal_speak(obj):
    obj.speak()

# Testing
a1 = Animal('jackie')
d1 = Dog('Lucy')

animal_speak(a1)  # The animal makes a sound
animal_speak(d1)  # The Lucy says woof!
```

## Polymorphism through abstract base classes

```python
## import some function and abstract method
## ABC is a abstract class
from abc import ABC, abstractmethod
```

```python
## abstract method are empty inside

from abc import ABC, abstractmethod

class Vehicle(ABC):
    def __init__(self, brand, type, model, year):
        self.brand = brand
        self.type = type
        self.model = model
        self.year = year

    @abstractmethod
    def drive(self):
        pass


class Car(Vehicle):
    def __init__(self, brand, type, model, year, num_doors):
        super().__init__(brand, type, model, year)
        self.num_doors = num_doors

    def drive(self):
        print("Driving a car")
        print(f"You are driving a {self.brand} {self.model} with {self.num_doors} doors")


class Motorbike(Vehicle):
    def __init__(self, brand, type, model, year, engine_size):
        super().__init__(brand, type, model, year)
        self.engine_size = engine_size

    def drive(self):
        print(f"You are driving a {self.brand} {self.model} with an engine size of {self.engine_size}")


# ✅ Correct instance creation
c1 = Car('Toyota', 'Sedan', 'GLI', '2020', 4)
c1.drive()
# Output:
# Driving a car
# You are driving a Toyota GLI with 4 doors

m1 = Motorbike('Honda', 'Sport', '125', '2015', '125CC')
m1.drive()
# Output:
# You are driving a Honda 125 with an engine size of 125CC
```

```python
c2 = Car('Honda', 'Sedan', 'City', 2015, 4)
c2.drive()

m2 = Motorbike('Kawasaki', 'Sport', 'CG', 2005, '150 CC')
m2.drive()
```
