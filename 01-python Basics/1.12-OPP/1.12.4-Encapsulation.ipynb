## Encapsulation in Python
- Encapsulation is a fundamental concept in object-oriented programming (OOP) that binds together the data.
 - It is a way to hide the internal state of an object from the outside world.

```python
## Public Variables 

class Person:
    def __init__(self, name, age, gender):
        self.name = name  ## Public Variales
        self.age = age    ## Public Variales
        self.gender = gender ## Public Variales
    
p1 = Person('Waseem', 27, 'Male')
print(p1.name) 
print(p1.age)   
print(p1.gender)  
```

```python
## Can print information from parents if inherint from child class
## Public variables show if the class is inherited from another class or external class

# protected variables cannot be accessed by external variables but inherited from parent to child class

## privated variables cannot be accessed by external variables and cannot be inherited from parent to child class
```

```python
## Now they are protected
class Person:
    def __init__(self, name, age, gender):
        self._name = name
        self._age = age
        self._gender = gender

p1 = Person("Waseem", 27, "Male")

## let do a class

class Employee(Person):
    def __init__(self, name, age, gender):
        super().__init__(name, age, gender)

e1 = Employee('Fakhar', 25, 'Male')

print(e1._name)
print(e1._age)
print(e1._gender)

## it shown after inherited but remember it will not work for private variables
```

```python
class Person:
    def __init__(self, name, age, gender):
        self.__name = name    # Correctly define private variables
        self.__age = age
        self.__gender = gender

## Getter and Setter Methods(instance or accessing variables)


        ## Getter Method
    def get_name(self):
        return self.__name

    def get_age(self):
        return self.__age

    def get_gender(self):
        return self.__gender

    # Setter Methods
    def set_name(self, name):
        self.__name = name

    def set_age(self, age):
        if age >= 0:
            self.__age = age
        else:
            print("Age should be greater than 0")

# Create object and test outside the class
p2 = Person('Fakhar', 22, 'male')

print(p2.get_name())
print(p2.get_age())

p2.set_age(25)
print(p2.get_age())

print(p2.get_gender())




    
```
