## Inheritance in Python
- Single Inheritance
- Multiple Inheritance

```python
## for class Use Captial letter
##  car is a blue print of data
class Car:
    def __init__(self, engine_type, model, year):
        self.engine_type = engine_type
        self.model = model
        self.year = year

    def drive(self):
        print(f"The {self.model} is driving")
```

```python
## Single Inheritance
## In single inheritance, a child class inherits the properties and methods of a single parent class.
##  super().__init__ is used for previous data

class Tata(Car):
    def __init__(self, model, year, color):
        super().__init__(self,model, year)
        self.color = color

    def start (self):
        print("f{self.model} car started")
    
t1 = Tata('A', 2025, 'Grey')

print(t1.color)
t1.start()
t1.drive()
```

```python
## Multiple Inheritance
##Multiple inheritance is a feature of object-oriented programming (OOP) that allows a class to inherit properties.

## class 1
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound")

## class 2

class pet:
    def __init__(self, owner):
        self.owner = owner


## class 3 - Child class

class Dog(Animal, pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        pet.__init__(self, owner)


d1 = Dog('Buddy', 'John')

print(d1.name)
d1.speak()
print(d1.owner)
```
