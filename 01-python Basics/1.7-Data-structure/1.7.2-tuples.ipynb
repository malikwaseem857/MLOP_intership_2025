## Tuples in Python
- Creating tuples
- Acessing tuples
- Methods
- Packing and Unpacking
- etc.

```python
## Tuples
# These are immutable , ordered collections of values that can be of any data type, including strings, integers, floats,

## list in [], tuples in ()
tpl = (1, "Waseem", 5.7, True, )
print(type(tpl))
```

```python
tpl = 1,2,3,4,5,6
print(type(tpl))
```

```python
lst = [1, 2 ,3 ,4 ,5, 'Waseem', 'Akram', True, False, 5.6, 5.7]

tpl = (tuple(lst)) ## Type casting
print(type(tpl))
```

```python
## Acessing Tuples
## Indexing

print(tpl[1])
print(tpl[9])

## Slicing Operations

print(tpl[0:])
print(tpl[:5])
print(tpl[2:7])
print(tpl[4::2])
```

```python
## Operation and Methods

print(tpl.count('Waseem'))
print(tpl.index('Waseem'))
```

```python
tpl * 3
```

```python
## Concartenation 
numbers = (1,2,3,4,5,6,7,8)

mixed_tpl = tpl + numbers

print(mixed_tpl)
```

```python
## Packing and Unpacking
## when we don't use () it is called packing 

tpl = 2,3,4,5,6,"waseem", "08", "Botanist"
```

```python
first, *middle, lst = tpl
```

```python
## Unpacking has 2-3 types
## Unpacking by esteric method
first, *middle, last = tpl

print(first)
print(middle)
print(last)  # Output: 1 ['2', '3']
```

```python
## 2nd type of Unpacking
tpl = "waseem", "Fakhar", "Ali"
a, b, c = tpl
print(a)
print(b)
print(c)
```

```python
## Iterating over tuples
## means we are looping 

tpl = 2,3,4,5,6,"waseem", "08", "Botanist"
for i in tpl:
        print(f"Hi,{i}")
```

```python
## Nested Tuples

nested_list = [[1,2,3,4,], ["Waseem", "Fakhar", "Ali"], [True, False, True, False]]

nested_tuple = ((1,2,3,4), ("Waseem", "Fakhar", "Ali"), (True, False, True, False))
```

```python
for i in nested_tuple:
    for j in i:
        print(f"{j*2}")
```

```python
## Tuples vs lists
nested_list = [[1,2,3,4,], ["Waseem", "Fakhar", "Ali"], [True, False, True, False]]

nested_tuple = ((1,2,3,4), ("Waseem", "Fakhar", "Ali"), (True, False, True, False))
```

```python
## Modifying the  list element
nested_list[0][2] = "PU"
print(nested_list)
```
