```python
# Creating List
#  if we run [], it will create a list with no elements
lst = []
print(type(lst))
```

```python
lst = [1,2,3,4,5,"Waseem", "Akram",'Malik','5.6','8.9','True','False']
print(lst)
```

```python
## How to access the list
## 1. Using index: list[0] or list[1]

print(lst[0])
print(lst[1])
print(lst[11])
print(lst[-1])
```

```python
## Slicing operation

lst[0:3]
```

```python
print(lst[0:3])
print(lst[2:6])
print(lst[2::2])
print(lst[2:])
print(lst[:7])
print(lst[-1:-10])
```

```python
print(lst[0:3])
print(lst[2:6])
print(lst[2::2])
print(lst[2:])
print(lst[:7])
print(lst[-1::-10])
```

```python
## Modifing the list elements

lst[7] = "Mircoinformatics"
print(lst)
```

```python
## list Methods  append means adding an element to the end of the list


lst.append('Waseem')
print(lst)  

out_put = lst.pop()
print(out_put)

lst.remove("Mircoinformatics")
print(lst)
```

```python
# sort in alphabatically
fruits = ['banana', 'apple', 'orange', 'grape','cat', 'dog']
fruits.sort()
print(fruits)
```

```python
## Arranged alphabatically and uppercase first
fruits = ['banana', 'apple', 'orange', 'grape','cat', 'dog', 'Waseem','Akram']
fruits.sort()
print(fruits)
```

```python
## Ilerating Over the list
fruits = ['banana', 'apple', 'orange', 'grape','cat', 'dog', 'Waseem','Akram']

for i in fruits:
    print(i.upper())
```

```python
fruits = ['banana', 'apple', 'orange', 'grape','cat', 'dog', 'Waseem','Akram']

for i in fruits:
    print(i.lower())
```

```python
fruits = ['banana', 'apple', 'orange', 'grape','cat', '', 'Waseem','Akram']

for i in fruits:
    print(i.lower())
```

```python
lst =[
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
num= []
for i in lst:
    num.append(i[0])

print(num)

```

```python
## Nested List
lst =[
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
num = []
for i in lst:
    for j in i:
        num.append(j)

print(num)
```

## list Comprehension

formate_1 : [*expression for item in iterable*]/

formate_2 : [*expression for item in iterable if condition*]

### Nested List Comprehension

formate_3: [*expression for iterable in variable if condition for iterable in condition2*]


```python
## List Comprehension

lst = []
for i in range(10):
    if i % 2 == 0:
        lst.append(i*i)
print(lst)
```

```python
## formate_2 : [*expression for item in iterable if condition*]
lst = [ i*i for i in range(10) if i % 2!= 0]
print(lst)
```

```python
lst =[
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
## nested list comprehension

nest_lst = [j for i in lst for j in i]

print(nest_lst)
```
