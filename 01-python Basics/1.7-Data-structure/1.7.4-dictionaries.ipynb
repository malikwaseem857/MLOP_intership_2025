## Dictionaries in Python
### A dictionary is a collection of key-value pairs. It is an unordered collection of items.


```python
## Creating Dictionaries

dict_1 = {}
print(type(dic_1))
```

```python
dict_1 = {"name": "Waseem", "age": 27, "city": "Lahore"}
print(type(dic_1))
```

```python
dict_1 = {"name": "Waseem", "age": 27, "city": "Lahore"}
print(dic_1)
```

## Some Important Points
### - key: Unique, Immutable


```python
## Accessing Data OTBO Keys
print(dict_1["name"])
print(dict_1["age"])
```

```python
dict_1 = {"name": "Waseem", "age": 27, "filed": "Plant Science", "name": "Haseeb"}

print(dic_1)
```

```python
## Dictionary Methods

## Acessong Data OTBO keys
print(dict_1["name"])
print(dict_1["age"])

## Accessing data dictionary methods

print(dic_1.get("name"))
print(dic_1.get("age"))
```

```python
## Other Methods

keys = dict_1.keys()
values = dict_1.values()
items = dict_1.items()

print(keys)
print(values)
print(items)
```

```python
## Shallow Copy
#Shallow copy is a process where a new object is created and the elements of the original object
## we can add, remove, replace the data in dictionaries
clone_1 = dict_1
print(clone_1)
```

```python
dict_1["name"] = "Waseem"

print(dict_1)
print(clone_1)
```

```python
clone_2 = dict_1.copy()

dict_1["name"] = "Fatima"

print(dict_1)
print(clone_2)
```

```python
## Updates, delete and adding new records

dict_1["age"] = 28 ## updating the exting age
print(dict_1)

dict_1["city"] = "Lahore" ## adding new recored, if any key is not exist, it will new created
print(dict_1)
del dict_1["city"]
print(dict_1)
```

```python
## Iterating over Dictionaries
#Dictionaries are iterable, meaning you can use a for loop to iterate over their keys, values.

for key in dict_1.keys():
    print(key)
```

```python
for value in dict_1.values():
    print(value)
```

```python
# for loops
for key,value in dict_1.items():
    print(f"{key}: {value}")
```

```python
nest_di = {
    '2022-ag-4002': {'name': 'Waseem', 'age': '27', 'Dep': 'BOT', 'Club': 'YPDC'},
    '2022-ag-4004': {'name': 'XYZ', 'age': '27', 'Dep': 'IOM', 'Club': 'YPDC'}
}

print(nest_di['2022-ag-4002'])
print(nest_di['2022-ag-4004'])
```

```python
## Iterating over Nested Dictionaries

for key,value in nest_di.items(): 
    for i, j in value.items():
        print(f"{key} > {i}: {j}")
```

```python
## how to find the frequency numbers

numbers = [1,1,1,2,2,2,3,3,3,4,4,4,5,5,5,6,6,6,7,7,7,9,9,0]

frequency = {}
for  i in numbers:
    if i in frequency:
        frequency[i] += 1
    else:
        frequency[i] = 1


print(frequency)
```
