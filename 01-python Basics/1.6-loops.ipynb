```python
## 1- For Loops 
# if i want to print each number in a list, it can be done using for loop.
# itribl
age = [12,21,14,15,31,21,22,22,24]
for i in age:
    print(i)
```

```python
# i%2==0: means even number it will show all even numbers
age = [12,21,14,15,31,21,22,22,24]

for i in age:
    if i%2==0:
        print(i)
```

```python
# it will print in rows
age = [12,21,14,15,31,21,22,22,24]

for i in age:
    if i%2==0:
        print(i, end=" ")
```

```python
for i in range(10):
    print(i, end=" ")
```

```python
for i in range(100):
    print(i, end=" ")
```

```python
# if i want odd numbers only
for i in range(100):
    if i%2!=0:
        print(i, end=" ")
```

```python
# this is a sum of numbers from 1 to 99
sum = 0
for i in range(100):
    sum = sum + i
print(sum)
```

```python
# Tricky Question same as above
sum = 0
for i in range(100):
    sum +=i
print(sum)
```

```python
# while loops works untill condition will remain true
num = 100
fact = 1
count = 1
while count<=num:
    fact = fact * count
    count = count + 1
print(fact)

```

```python
# it happen it oppistie direction
num = 1
fact = 1
count = 100
while count>=1:
    fact = fact * count
    count = count - 1
print(fact)
```

## Loops Control Statments

```python
## pass Statements
for i in range(1, 100):
    pass
```

```python
## Continue

for i in range(10):
    if i % 2 == 0:
        continue
    print(i, end=" ")
```

```python
## Break
for i in range(50):
    if i==38:
        break
    print(i, end=" ")
```

```python
## Nested Loops
#A nested loop is a loop that is contained within another loop. The outer loop iterates over th

names = ['Waseem', 'Akram', 'Malik']
for i in range(3):
    for j in names:
        print(f"{i} {j}")
```
