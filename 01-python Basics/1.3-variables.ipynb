```python
#  a has a value it will be called variable in python
# String always used in ""
a= 23 # int
name= "Waseem" # str
height= 5.7    # float
is_adult= True # boolean

```

```python
print(a)
print(name)
print(height)
print(is_adult)
```

```python
print(type(a))
print(type(name))
print(type(height))
print(type(is_adult))
```

```python
## python is a dynamically typed languange

var = 24
print(type(var))
var = "Waseem"
print(type(var))
var = height
print(type(var))
var = 5.7
print(type(var))
```

```python
# Type Conversion we changed str into int
# Str number changed to number
# str letters cannot changes to number
age = '27'
print(type(age))
age_1= int(age)
print(type(age_1))
```

```python
# Naming convention
# always use smallcase letter or numbers
# always start with _start, start_
# not used special characters or number at start
```

```python
## Input Function used to take input from users
num_1 = input("Enter a number")
num_2 = input("Enter a second number")
```

```python
print(num_1)
print(num_2)
```

```python
# we changed the str into int
num_1 = int(input("Enter a number"))
num_2 = int(input("Enter a second number"))
```

```python
print(type(num_1))
print(type(num_2))
```

```python
# Simple calculator that we made
num_1 = int(input("Enter a number"))
num_2 = int(input("Enter a second number"))

sum = num_1 + num_2
diff = num_2 - num_1
mult = num_1 * num_2
div = num_1 / num_2

print(sum)
print(diff)
print(mult)
print(div)
```
