## File Operation in Python
#### Write Mode (w)
#### Append Mode (a)
#### Read and Write Mode (r+)
#### Append and Write Mode (a+)
#### Write byte Mode (wb)
#### Read byte Mode (rb)
#### Append byte Mode (ab)
#### Read and Write byte Mode (r+b)



```python
## Specific formate
with open('example.txt', 'r' ) as file:
    content = file .read()

print(content)
```

```python
## Overriddding the existing

with open('example.txt', 'w') as file:
    file.write ('Waseem is learning data science.\n')
    file .write ('He is a python learner.')
    print('the file has been written')
```

```python
## Adding the content to the existing file
## /n   means new line argument

with open('example.txt', 'a') as file:
    file.write(' This is the new content.\n')

with open('example.txt', 'r') as file:
    print(file.read())  # Output: This is the new content.
```

```python
## write a binary file using python

data = b'\x00\x01\x02'
with open('binary_file.bin', 'wb') as file:
    file.write(data)
print('file has been generated')
```

```python

with open('binary_file.bin', 'rb') as file:
    content = file.read()
    print(content)
```
